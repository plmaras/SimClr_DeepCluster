```python
def get_color_distortion(s=1.0):
    color_jitter = transforms.ColorJitter(0.8*s, 0.8*s, 0.8*s, 0.2*s)
    rnd_color_jitter = transforms.RandomApply([color_jitter], p=0.8)
    rnd_gray = transforms.RandomGrayscale(p=0.2)
    color_distort = transforms.Compose([rnd_color_jitter,rnd_gray])
    return color_distort
```

```python
RRezCropFlip = torchvision.transforms.Compose([transforms.RandomResizedCrop(size=(32,32)),
                                               transforms.RandomHorizontalFlip(p=0.5)])
```

```python
class Model(nn.Module):

    def __init__(self,model,hidden_dim,out_dim=128,rep_lay = -1):
        super().__init__()
        # self.model = nn.Sequential(*list(model.children())[:rep_lay])
        self.model = model
        self.replay = rep_lay
        self.out_dim = out_dim
        self.hidden_dim = hidden_dim
        self.ln2 = nn.Linear(hidden_dim,out_dim)
        self.sgn = nn.Sigmoid()
        
    def forward(self,x):
        rep = self.model(x)
        rep = torch.flatten(rep,start_dim = 1)
        h = nn.Linear(rep.shape[1],self.hidden_dim)(rep)
        z = self.ln2(self.sgn(h))
        return z

    def getrep(self,x):
        rep = self.model(x)
        rep = torch.flatten(rep,startdim = 1)
        return rep   
```
